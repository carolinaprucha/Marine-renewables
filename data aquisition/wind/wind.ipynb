```python
import cdsapi

c = cdsapi.Client(
    url="https://cds.climate.copernicus.eu/api",
    key="1a6d4adf-df37-457d-a4c2-56ca4ff23470"
)

# Download year by year to keep requests small
import os

output_dir = r'C:\Users\Carolina\Documents\MSc Civil Engineering\Marine renewables\marine renewables\Marine-renewables\data aquisition\wind_data'

for year in range(2010, 2021):
    output_file = os.path.join(output_dir, f'era5_wind_{year}.nc')
    
    if os.path.exists(output_file):
        print(f"{year} already downloaded, skipping...")
        continue
    
    print(f"Downloading {year}...")
    
    c.retrieve(
        'reanalysis-era5-single-levels',
        {
            'product_type': ['reanalysis'],
            'variable': [
                '10m_u_component_of_wind',
                '10m_v_component_of_wind',
                '100m_u_component_of_wind',
                '100m_v_component_of_wind',
            ],
            'year': [str(year)],
            'month': [f'{m:02d}' for m in range(1, 13)],
            'day': [f'{d:02d}' for d in range(1, 32)],
            'time': ['00:00', '03:00', '06:00', '09:00', '12:00', '15:00', '18:00', '21:00'],
            #'time': [f'{h:02d}:00' for h in range(0, 24)],
            'area': [37.75, -9.02, 37.74, -9.01],  # tiny box = single grid point at Sines
            'data_format': 'netcdf',
            'download_format': 'unarchived',
        },
        output_file
    )
    print(f"{year} done!")

print("All years downloaded!")
```
