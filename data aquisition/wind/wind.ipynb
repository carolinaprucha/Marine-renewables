```python
import cdsapi
import xarray as xr
import numpy as np

# ── Download ERA5 wind data for Sines ─────────────────────────────
c = cdsapi.Client()

c.retrieve(
    'reanalysis-era5-single-levels',
    {
        'product_type': 'reanalysis',
        'variable': [
            '10m_u_component_of_wind',   # wind speed East-West at 10m
            '10m_v_component_of_wind',   # wind speed North-South at 10m
            '100m_u_component_of_wind',  # wind speed East-West at 100m
            '100m_v_component_of_wind',  # wind speed North-South at 100m
        ],
        'year': [str(y) for y in range(2010, 2021)],  # 2010 to 2020, matching ECHOWAVE        'month': [f'{m:02d}' for m in range(1, 13)],
        'day': [f'{d:02d}' for d in range(1, 32)],
        'time': [f'{h:02d}:00' for h in range(0, 24)],
        'area': [38.5, -9.8, 37.0, -8.2],  # N, W, S, E — box around Sines
        'format'
        : 'netcdf',
    },
    r'C:\Users\Carolina\Documents\MSc Civil Engineering\Marine renewables\marine renewables\Marine-renewables\data aquisition\wind_data\era5_wind_sines.nc'
)

print("Download complete!")
```
